```python
#1 carteira de motorista

print("Você tem idade pra tirar a carteira de motorista")
a = int(input("Qual a sua idade? "))
b = 18 - a
if a < 18:
  print(f'Você não tem idade para tirar sua carteira de motorista. Você poderá tirá-la em aproximadamente {b} anos.')
else:
  print(f'Você tem idade pra tirar a carteira de motorista')
```

```python
#2 Par ou Impar

a = float(input("Digite um número, e direi se é par ou ímpar: "))
if a %2 == 0:
  print(f'{a} é um número par')
else:
  print(f'{a} é um número ímpar')
```

```python
#3 Maior, menor ou igual
print("Digite dois números, e direi qual o maior, e qual o menor")
a = float(input("Digite número: "))
b = float(input("Digite outro número: "))

if a < b:
  print(f'{a} é menor que {b}.')
elif a > b:
  print(f'{a} e maior que {b}.')
else:
  print(f'{a} é igual a {b}')
```

```python
#4 Situação do aluno

a = float(input("Digite sua nota final: "))

if a >= 7:
  print("Você foi aprovado!")
elif a < 5:
  print("Você foi reprovado.")
else:
  print("Você está em recuperação...")
```

```python
#5 Desconto na compra

a = float(input("Digite o valor da sua compra: "))

if a >= 100:
  b = a / 10
  c = a - b
  print(f"Sua compra ficará {c} reais, após aplicar 10% de desconto.")
else:
  print(f"Sua compra ficará {a} reais, sem desconto.")
```

```python
#6 Positivo, Negativo ou Zero

a = float(input("Digite um número: "))

if a > 0:
  print(f"{a} é positivo")
elif a < 0:
  print(f"{a} é negativo")
else:
  print(f"{a} é igual a zero")
```

```python
#7 Idade

a = int(input("Digite sua idade: " ))

if a < 12:
  print("Você é criança")
elif a >= 60:
  print("Você é idoso")
elif a >= 12 and a < 18:
  print("Você é adolescente")
else:
  print("Você é adulto")
```

```python
#8 Situação do aluno pt2

a = float(input("Digite sua nota: "))
b = float(input("Digite sua segunda nota: "))
c = (a + b) / 2

if c >= 7:
  print("Você foi aprovado!")
elif c < 5:
  print("Você foi reprovado.")
else:
  print("Você está em recuperação...")
```

```python
#10 IMC

p = float(input("Digite seu peso: "))
a = float(input("Digite sua altura: "))

imc = p / (a * a)

if imc < 18.5:
  print("Abaixo do peso")
elif imc >= 30:
  print("Obesidade")
elif imc >= 25 and imc < 30:
  print("Sobrepreso")
else:
  print("Peso normal")
```
