```python
# Cofre digital

# UM cofre possui um código de 5 números inteiros

# Desenvolva um programa que:

# Leia os 5 números utilizando um laço de repetição

# Ao final informar:

# 1 - Quantos números são pares
# 2 - Quantos números são ímpares
# 3 - O maior número
# 4 - A soma de todos os números
# 5 - A média dos números
# 6 - O menor número
# 7 - Quantos números são maiores do que a média

pares = 0
impares = 0
soma = 0

num = int(input("Digite o 1º número: "))
maior = num
menor = num
soma += num

if num % 2 == 0:
    pares += 1
else:
    impares += 1

numeros = [num]

for i in range(2, 6):
    num = int(input(f"Digite o {i}º número: "))
    numeros.append(num)
    soma += num

    if num % 2 == 0:
        pares += 1
    else:
        impares += 1

    if num > maior:
        maior = num

    if num < menor:
        menor = num

media = soma / 5

maiores_media = 0
for num in numeros:
    if num > media:
        maiores_media += 1

print("\nResultado:")
print("1 - Quantidade de números pares:", pares)
print("2 - Quantidade de números ímpares:", impares)
print("3 - Maior número:", maior)
print("4 - Soma dos números:", soma)
print("5 - Média dos números:", media)
print("6 - Menor número:", menor)
print("7 - Quantos números são maiores que a média:", maiores_media)
```

```python
# Número Palíndromo

# Escolha um número inteiro positivo
# Inverta ele
# Informe se é um palíndromo


numero = int(input("Digite um número inteiro positivo: "))

invertido = int(str(numero)[::-1])

if numero == invertido:
    print(f"{numero} é um número palíndromo.")
else:
    print(f"{numero} não é um número palíndromo.")
```
